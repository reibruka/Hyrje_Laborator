## 1. Çfarë është Miniconda dhe Pse Përdoret?

**Miniconda** është një instalues minimal dhe i thjeshtësuar për Conda. Ndryshe nga Anaconda (e cila vjen e ngarkuar me qindra paketa), Miniconda përfshin vetëm elementët bazë: Conda, Python, paketata nga të cilat ata varen, dhe disa mjete thelbësore si `pip`.

### Pse përdoren mjediset (environments) e Conda?
* **Izolim Total:** Shmang konfliktet e versioneve midis projekteve të ndryshme duke i mbajtur paketave të ndara në mjedise specifike[cite: 1].
* **Fleksibilitet:** Mundëson instalimin e versioneve të ndryshme të Python (p.sh., Python 3.8 për një projekt dhe Python 3.11 për një tjetër) në të njëjtën makinë[cite: 1].
* **Riprodhueshmëri:** Mund t'i ndani lehtësisht mjediset tuaja me të tjerët përmes skedarëve të konfigurimit, duke siguruar që kodi të funksionojë njësoj kudo[cite: 1].
* **Ndër-platformë (Cross-Platform):** Conda menaxhon librari dhe binarë që shkojnë përtej Python, gjë që është jetike për Data Science dhe Machine Learning[cite: 1].

## 2. Konceptet Bazë

* **Conda:** Sistemi që menaxhon paketat dhe mjediset virtuale në Windows, macOS dhe Linux[cite: 1].
* **Environment (Mjedisi):** Një direktori e izoluar ku instalohet një grup specifik paketash pa ndikuar në sistemin global[cite: 1].
* **Channel (Kanali):** Lokacionet ose adresat URL ku Conda kërkon për të shkarkuar paketat (si p.sh. `defaults` ose `conda-forge`)[cite: 1].
