## 1. Pse përdorim Python për Analizën e të Dhënave?

Python është bërë gjuha standarde në shkencën e të dhënave (Data Science) dhe laboratorët shkencorë për disa arsye kryesore:
* **Sintaksë e thjeshtë:** Është i lehtë për t'u lexuar dhe shkruar, duke lejuar që fokusimi të jetë te problemi shkencor dhe jo te kompleksiteti i kodit.
* **Ekosistem i fuqishëm:** Ka librari të jashtëzakonshme si `NumPy` për llogaritje matematike, `Matplotlib` për grafikë, dhe `SciPy` për analiza shkencore.
* **Komunitet i madh:** Çdo problem apo gabim që mund të hasim ka pothuajse gjithmonë një zgjidhje të gatshme online.

## 2. Variablat dhe Array-t (Matricat)

* **Variablat:** Përdoren për të ruajtur vlera të thjeshta (numra, tekste).
* **Array (Vektorët/Matricat):** Libraria `NumPy` na lejon të krijojmë struktura të quajtura `ndarray`. Ndryshe nga listat e thjeshta të Python, array-t e NumPy lejojnë veprime matematike të shpejta dhe efikase mbi të gjithë vektorin njëkohësisht (vektorizim).

```python
import numpy as np

# Shembull me Variabla
pi_vlera = 3.14159
print(f"Vlera e pi-së: {pi_vlera}")

# Krijimi i një Array me NumPy (p.sh. boshti X nga 0 në 10)
x = np.linspace(0, 10, 20)
print("Array X:", x)
```

## 3. Krijimi i Grafikëve dhe Vizualizimi i të Dhënave

Për të krijuar grafikë profesionalë përdorim librarinë `matplotlib.pyplot`. Është e rëndësishme që çdo grafik të ketë emërtim të boshteve, titull dhe legjendë nëse ka më shumë se një vijë.

```python
import matplotlib.pyplot as plt

# Gjenerojmë të dhëna fiktive me pak zhurmë (Noise) si në laborator
np.random.seed(42)
y_te_dhena = 2 * x + 1 + np.random.normal(0, 1.5, len(x))

# Ndërtimi i grafikut
plt.figure(figsize=(8, 5))
plt.scatter(x, y_te_dhena, color='red', label='Të dhënat eksperimentale (me zhurmë)')
plt.xlabel('Boshti X (Njësitë)')
plt.ylabel('Boshti Y (Njësitë)')
plt.title('Grafiku i të Dhënave Eksperimentale')
plt.legend()
plt.grid(True)
plt.show()
```

## 4. Përshtatja e të Dhënave (Data Fitting)

Në laborator, shpesh na duhet të gjejmë ekuacionin matematik që përshkruan më mirë pikat tona eksperimentale. Ky proces quhet **Data Fitting**. Për një vijë të drejtë ($y = mx + c$), mund të përdorim funksionin `curve_fit` nga libraria `SciPy`.

```python
from scipy.optimize import curve_fit

# Përcaktojmë funksionin linear model
def model_linear(x, m, c):
    return m * x + c

# Kryejmë procesin e Fitting
paramatrat, kovarianca = curve_fit(model_linear, x, y_te_dhena)
m_gjetur, c_gjetur = paramatrat

print(f"Ekuacioni i gjetur: Y = {m_gjetur:.2f}*X + {c_gjetur:.2f}")
```

## 5. Grafiku Final (Të dhënat + Vija e Fitting)

Këtu bashkojmë pikat tona eksperimentale me vijën teorike që llogaritëm nga procesi i fitting.

```python
plt.figure(figsize=(8, 5))
plt.scatter(x, y_te_dhena, color='red', label='Pikat Eksperimentale')
plt.plot(x, model_linear(x, m_gjetur, c_gjetur), color='blue', linewidth=2, label=f'Best Fit: y={m_gjetur:.2f}x+{c_gjetur:.2f}')

plt.xlabel('Boshti X')
plt.ylabel('Boshti Y')
plt.title('Grafiku Final me Data Fitting')
plt.legend()
plt.grid(True)
plt.savefig('figs/grafiku_final.png') # Ruhet automatikisht te folderi figs
plt.show()
```

## 6. Fluksi Tipik i Punës në Analizën e të Dhënave (Workflow)

Një proces tipik i analizës në laborator ndjek këto hapa kronologjikë:
1. **Mbledhja/Importimi:** Leximi i të dhënave nga skedarët e jashtëm (p.sh. skedarë `.txt` ose `.csv` të ruajtur te folderi `data/`).
2. **Pastrimi dhe Inspektimi:** Kontrolli i të dhënave për vlera të gabuara ose të habitshme dhe vendosja e tyre në array-t e NumPy.
3. **Vizualizimi Fillestar:** Ndërtimi i një grafiku të thjeshtë me pika (Scatter Plot) për të kuptuar sjelljen e fenomenit.
4. **Modelimi Matematik (Fitting):** Zgjedhja e një ekuacioni teorik dhe përshtatja e parametrave me të dhënat.
5. **Eksportimi dhe Raportimi:** Ruajtja e grafikëve finalë te folderi `figs/` dhe nxjerrja e konkluzioneve në raport.
