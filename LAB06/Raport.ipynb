## 1. Konceptet Bazë të Matjes

Matja është procesi i krahasimit të një madhësie fizike të panjohur me një madhësi standarde të përcaktuar mirë, e cila quhet njësi matëse. Në këtë laborator fokusohemi në tri madhësitë fundamentale:
* **Gjatësia ($L$):** Distanca hapësinore midis dy pikave. (Njësia SI: metër, $m$)
* **Masa ($m$):** Sasia e lëndës që përmban një trup dhe masa e inertesisë së tij. (Njësia SI: kilogram, $kg$)
* **Koha ($t$):** Kohëzgjatja ose intervali midis dy ngjarjeve. (Njësia SI: sekondë, $s$)

### Saktësia (Precision) vs. Rezolucioni (Resolution)
* **Rezolucioni:** Ndarja më e vogël që mund të lexohet drejtpërdrejt në shkallën e një pajisjeje matëse (p.sh., rezolucioni i një liniari të thjeshtë është $1\text{ mm}$).
* **Saktësia (Precision):** Shkalla e përsëritshmërisë së matjeve. Nëse e masim të njëjtin objekt disa herë dhe marrim vlera shumë të afërta me njëra-tjetrën, matja është e saktë.

## 2. Pajisjet Matëse dhe Procedura

| Madhësia | Pajisja Matëse | Rezolucioni Tipik | Procedura e Matjes |
| :--- | :--- | :--- | :--- |
| **Gjatësi** | Nonius (Caliper) / Mikrometr | $0.05\text{ mm}$ / $0.01\text{ mm}$ | Përputhja e nofullave me objektin, leximi i shkallës kryesore dhe asaj nonius. |
| **Masë** | Peshore Elektronike | $0.01\text{ g}$ | Tarimi (Zeroing) i peshores, vendosja e objektit në qendër pa lëkundje. |
| **Kohë** | Kronometër Dixhital | $0.01\text{ s}$ | Nisja dhe ndalimi sinkron me fillimin dhe mbarimin e fenomenit fizik. |

## 3. Formulat Kryesore për Analizën e të Dhënave

Kur kryejmë disa matje të përsëritura ($x_1, x_2, ..., x_n$), përdorim këto formula:

1. **Vlera Mesatare ($\\bar{x}$):**
$$\\bar{x} = \\frac{1}{n} \\sum_{i=1}^{n} x_i$$

2. **Gabimi Absolut ($\\Delta x$):** Për një matje të vetme, shpesh merret sa gjysma e rezolucionit të pajisjes, ose përdoret devijimi standard për matje të përsëritura.

3. **Gabimi Relativ ($epsilon$):**
$$\\epsilon = \\frac{\\Delta x}{\\bar{x}} \\times 100\%$$

```python
import numpy as np

# Të dhëna fiktive nga matja e gjatësisë së një objekti (në mm)
matjet_gjatesi = np.array([25.4, 25.5, 25.3, 25.4, 25.6])

# Llogaritja e vlerës mesatare
vlera_mesatare = np.mean(matjet_gjatesi)

# Gabimi absolut (Devijimi Standard i thjeshtuar)
gabimi_absolut = np.std(matjet_gjatesi, ddof=1)

# Gabimi relativ në përqindje
gabimi_relativ = (gabimi_absolut / vlera_mesatare) * 100

print(f"Vlera Mesatare: {vlera_mesatare:.2f} mm")
print(f"Gabimi Absolut: ±{gabimi_absolut:.2f} mm")
print(f"Gabimi Relativ: {gabimi_relativ:.2f}%")
```

## 4. Gabimet Tipike në Laborator (Problemet)

Gjatë matjeve shfaqen gjithmonë gabime, të cilat i klasifikojmë në:
* **Gabime Sistematike:** Ndodhin për shkak të pajisjeve të pakalibruara (p.sh. një kronometër që ecën ngadalë ose peshore që nuk niset nga zeroja). Këto gabime zhvendosin të gjitha matjet në një drejtim.
* **Gabime Rastësore:** Ndodhin nga ndryshimet e paparashikueshme mjedisore ose kufizimet njerëzore (p.sh. koha e refleksit gjatë shtypjes së kronometrit). Mund të zvogëlohen duke bërë shumë matje dhe duke nxjerrë mesataren.
* **Gabimi i Paralaksit:** Ndodh kur syri i vëzhguesit nuk është pingul me shkallën matëse, duke lexuar një vlerë të gabuar.

## 5. Rezultatet dhe Diskutimi

Nga përpunimi i të dhënave të mbledhura në kodin e mësipërm, vëmë re se gabimi relativ është i ulët, gjë që tregon për një saktësi të lartë gjatë eksperimentit. 

Për të minimizuar gabimet në laboratorët e ardhshëm, duhet:
1. Të bëhet gjithmonë kalibrimi ose 'tarimi' i peshoreve përpara çdo matjeje.
2. Të shmanget gabimi i paralaksit duke parë direkt në mënyrë vertikale mbi shkallët me vija.
3. Të rritet numri i matjeve të përsëritura për të zvogëluar ndikimin e gabimeve rastësore.
