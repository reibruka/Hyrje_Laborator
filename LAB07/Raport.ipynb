## 1. Matjet Direkte vs. Indirekte dhe Dendësia

* **Matje Direkte:** Është procesi ku vlera e një madhësie fizike merret drejtpërdrejt nga pajisja matëse (p.sh. matja e masës me peshore ose e gjatësisë me kaliper).
* **Matje Indirekte:** Ndodh kur madhësia fizike nuk mund të matet direkt, por llogaritet përmes një formule matematike duke përdorur madhësi të tjera që janë matur direkt.

### Dendësia ($\rho$)
Dendësia është një shembull tipik i matjes indirekte. Ajo përcaktohet si masa për njësi të vëllimit:
$$\rho = \frac{m}{V}$$
Ku masa ($m$) matet direkt, ndërsa vëllimi ($V$) llogaritet në mënyrë indirekte duke matur dimensionet gjeometrike të objektit.

## 2. Pse nevojitet përhapja e pasigurisë?

Çdo matje direkte ka një **pasiguri absolute** (gabim) të vetën, e cila vjen nga rezolucioni i pajisjes apo gabimet njerëzore (shënohet si $\Delta m$ ose $\Delta x$). 

Kur këto madhësi i vendosim në një formulë (si ajo e dendësisë), pasiguritë e tyre fillestare "përhapen" dhe ndikojnë në rezultatin final. **Përhapja e pasigurisë** na ndihmon të llogarisim me saktësi se sa i sigurt është rezultati ynë indirekt bazuar tek gabimet e matjeve direkte.

## 3. Pajisjet dhe Objektet e Eksperimentit

### Pajisjet Matëse:
* **Peshore Elektronike:** Përdoret për matjen direkte të masës ($m$) me pasiguri instrumentale $\Delta m$.
* **Vizore, Kaliper (Nonius) ose Mikrometër:** Përdoren për matjen e dimensioneve lineare me saktësi të lartë.

### Objektet me formë të rregullt gjeometrike:
1. **Kub / Paralelepiped:** Me dimensione $a, b, c$. Vëllimi: $V = a \cdot b \cdot c$
2. **Cilindër:** Me rreze $r$ (ose diametër $d$) dhe lartësi $h$. Vëllimi: $V = \pi r^2 h$
3. **Sferë:** Me rreze $r$. Vëllimi: $V = \frac{4}{3}\pi r^3$

## 4. Modeli Fizik dhe Përhapja e Pasigurisë (Rasti i Cilindrit)

Le të marrim si model një **cilindër**. Vëllimi i tij është $V = \pi r^2 h$. 
Nëse përdorim diametrin $d = 2r$, formula bëhet:
$$V = \frac{\pi d^2 h}{4}$$

Sipas rregullave të analizës matematike (derivatet pjesore), pasiguria relative e vëllimit gjehet si:
$$\frac{\Delta V}{V} = \sqrt{\left(2\frac{\Delta d}{d}\right)^2 + \left(\frac{\Delta h}{h}\right)^2}$$

Më pas, pasi gjejmë $\Delta V$, pasiguria absolute e dendësisë ($\Delta \rho$) llogaritet nga formula:
$$\frac{\Delta \rho}{\rho} = \sqrt{\left(\frac{\Delta m}{m}\right)^2 + \left(\frac{\Delta V}{V}\right)^2}$$
$$\Delta \rho = \rho \cdot \sqrt{\left(\frac{\Delta m}{m}\right)^2 + \left(\frac{\Delta V}{V}\right)^2}$$

## 6. Analiza e të Dhënave me Python

Në këtë pjesë do të përdorim një skript Python për të llogaritur vëllimin, dendësinë dhe përhapjen e saktë të pasigurisë.

```python
import numpy as np

# 1. Të dhënat e matura direkt (Shembull për një cilindër bakri)
m = 150.4        # në gram (g)
Delta_m = 0.01   # pasiguria e peshores elektronike

d = 20.2         # diametri në mm
Delta_d = 0.05   # pasiguria e kaliperit

h = 52.1         # lartësia në mm
Delta_h = 0.05   # pasiguria e kaliperit

# Kovertimi i dimensioneve në cm që dendësia të dalë në g/cm3
d_cm = d / 10
h_cm = h / 10
Delta_d_cm = Delta_d / 10
Delta_h_cm = Delta_h / 10

# 2. Llogaritja e Vëllimit (V) dhe Pasigurisë së tij (Delta_V)
V = (np.pi * (d_cm**2) * h_cm) / 4
pasiguria_relative_V = np.sqrt((2 * Delta_d_cm / d_cm)**2 + (Delta_h_cm / h_cm)**2)
Delta_V = V * pasiguria_relative_V

# 3. Llogaritja e Dendësisë (rho) dhe Pasigurisë së saj (Delta_rho)
rho = m / V
pasiguria_relative_rho = np.sqrt((Delta_m / m)**2 + (Delta_V / V)**2)
Delta_rho = rho * pasiguria_relative_rho

print(f"Vëllimi i llogaritur: {V:.3f} ± {Delta_V:.3f} cm³")
print(f"Dendësia e llogaritur: {rho:.3f} ± {Delta_rho:.3f} g/cm³")
```
