## 1. Qëllimi i Laboratorit dhe Natyra Statistikore

Qëllimi i këtij laboratori është të tregojë se rezultatet eksperimentale në mjedise reale nuk përshkruhen gjithmonë nga një vlerë e vetme apo konstante, por nga një **shpërndarje probabilitare**.

Një shembull perfekt i kësaj është **koha e reagimit njerëzor**. Ajo është një madhësi statistikore që ndryshon nga matja në matje për shkak të faktorëve të panumërt biologjikë dhe mjedisorë. Rezultati i eksperimentit nuk është një numër i vetëm, por një bashkësi vlerash (data set) që grumbullohet rreth një vlere qendrore duke formuar një model shpërndarjeje.

## 2. Metoda Eksperimentale (Rënia e Vizores)

Për të matur kohën e reagimit pa pajisje të ndërlikuara elektronike, përdoret metoda klasike e rënies së vizores:
1. **Pozicionimi:** Eksperimentuesi i parë mban një vizore vertikalisht nga skaji i sipërm.
2. **Gati:** Eksperimentuesi i dytë (subjekti) mban gishtat e tij në nivelin e vijës zero të vizores, pa e prekur atë.
3. **Lëshimi:** Pa paralajmërim, vizorja lihet të bjerë e lirë. Subjekti duhet ta kapë atë sa më shpejt të jetë e mundur.
4. **Leximi:** Shënohet distanca e rënies ($d$) në centimetra direkt nga pika ku u kap vizorja.

### Modeli Fizik
Nga ligji i rënies së lirë ($d = \frac{1}{2}gt^2$), koha e reagimit ($t$) llogaritet në mënyrë indirekte si:
$$t = \sqrt{\frac{2d}{g}}$$
Ku $g \approx 9.81 \text{ m/s}^2$ (ose $981 \text{ cm/s}^2$).

## 3. Mbledhja, Organizimi i të Dhënave dhe Madhësitë Statistikore

Për të ndërtuar një profil statistikor të vlefshëm, procedura e mësipërme përsëritet një numër të madh herësh (p.sh., 50 deri në 100 herë). Të dhënat organizohen në një array në Python për të llogaritur parametrat kyç:
* **Vlera Mesatare (Mean - $\mu$):** Qendra e shpërndarjes.
* **Devijimi Standard (Standard Deviation - $\sigma$):** Mat shpërndarjen ose gjerësinë e të dhënave rreth vlerës mesatare.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Simulimi i të dhënave të mbledhura nga 50 matje të distancës (në cm)
np.random.seed(17)
distancat_cm = np.random.normal(loc=18.0, scale=3.5, size=50)
distancat_cm = np.clip(distancat_cm, 5.0, 30.0) # Kufizimi brenda vizores

# 2. Konvertimi i distancave në sekonda (Koha e Reagimit)
g = 981.0 # cm/s^2
kohet_s = np.sqrt((2 * distancat_cm) / g)

# 3. Llogaritja e madhësive statistikore
mesatarja = np.mean(kohet_s)
devijimi_std = np.std(kohet_s, ddof=1)

print(f"Numri i matjeve: {len(kohet_s)}")
print(f"Koha Mesatare e Reagimit: {mesatarja:.3f} sekonda")
print(f"Devijimi Standard: ±{devijimi_std:.3f} sekonda")
```

## 4. Histogrami dhe Normalizimi

Për të parë se si janë përhapur vlerat tona, ndërtojmë një **Histogram**. 
Për të bërë krahasimin me një model teorik probabilitar, histogrami duhet të **normalizohet** (duke vendosur parametrin `density=True`). Kjo bën që sipërfaqja totale e kolonave të histogramit të jetë e barabartë me 1, duke u kthyer në një Shpërndarje të Densitetit të Probabilitetit (PDF).

```python
# Ndërtimi i grafikut
plt.figure(figsize=(9, 5))

# Histogrami i normalizuar i të dhënave eksperimentale
count, bins, ignored = plt.hist(kohet_s, bins=10, density=True, alpha=0.6, color='g', edgecolor='black', label='Të dhënat Eksperimentale')

# 5. Krahasimi me Shpërndarjen Normale Teorike (Kurba Gaussian)
xmin, xmax = plt.xlim()
x_boshti = np.linspace(xmin, xmax, 100)
kurba_teorike = norm.pdf(x_boshti, mesatarja, devijimi_std)
plt.plot(x_boshti, kurba_teorike, 'r-', linewidth=2, label=f'Kurba Normale Teorike\n(\u03bc={mesatarja:.3f}s, \u03c3={devijimi_std:.3f}s)')

# Elementët e grafikut
plt.xlabel('Koha e Reagimit (sekonda)')
plt.ylabel('Densiteti i Probabilitetit')
plt.title('Histogrami i Normalizuar dhe Krahasimi me Shpërndarjen Normale')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

# Ruajtja e grafikut në folderin figs
plt.savefig('figs/histogrami_reagimit.png')
plt.show()
```

## 5. Gabimet e Zakonshme (Gabimet)

Gjatë këtij eksperimenti, disa gabime tipike që mund të dëmtojnë formën e pastër të shpërndarjes janë:
* **Paralajmërimi psikologjik:** Nëse personi që lëshon vizoren bën një lëvizje treguese ose numëron mbrapsht, subjekti reagon më shpejt se koha e tij reale (gabim sistematik).
* **Prekja e parakohshme:** Ndodh kur gishtat mbahen shumë afër ose prekin lehtë vizoren përpara lëshimit, duke krijuar fërkim dhe duke rritur artificialisht kohën e regjistruar.
* **Lodhja ose Mësimi:** Nëse matjet zgjasin shumë, subjekti mund të lodhet (duke rritur kohën e reagimit) ose të fillojë të mësohet me ritmin (duke e zvogëluar atë). Kjo bën që shpërndarja të mos jetë krejtësisht simetrike.
