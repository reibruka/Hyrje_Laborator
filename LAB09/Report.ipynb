## 1. Koncepti i Kohës së Ekzekutimit

Koha e ekzekutimit është koha reale që i duhet një programi apo kodi kompjuterik për të përfunduar një detyrë të caktuar. Kjo madhësi matet zakonisht në **sekonda (s)** ose **milisekonda (ms)**.

Koha e ekzekutimit të një programi nuk është fikse, por varet drejtpërdrejt nga disa faktorë kritikë:
1. **Algoritmi i përdorur:** Mënyra dhe hapat logjikë se si zgjidhet problemi (p.sh., kërkimi linear vs. kërkimi binar).
2. **Madhësia e input-it ($n$):** Sasia e të dhënave që algoritmi duhet të përpunojë.
3. **Hardueri dhe Sistemi:** Fuqia llogaritëse e procesorit (CPU), shpejtësia e memories (RAM) dhe menaxhimi i Burimeve nga Sistemi Operativ (OS).

### Fokusimi i Laboratorit
Në këtë eksperiment, ne do të fokusohemi specifikisht në studimin e lidhjes grafike dhe funksionale midis:
$$\text{Koha e ekzekutimit } T(n) \quad \text{vs.} \quad \text{Madhësia e input-it } n$$

**Ideja kryesore:** Sa më shumë të dhëna të përpunojë një algoritëm, aq më shumë hapa llogaritës dhe rrjedhimisht më shumë kohë kërkon për të përfunduar.

## 2. Pajisjet dhe Mjetet

Për këtë eksperiment nuk përdorim mjete fizike mekanike, por një mjedis kompjuterik modelimi:
* **Kompjuter** i pajisur me mjedisin Python (përmes Jupyter Notebook).
* **Libraritë standarde të Python:**
  * `time` ose `timeit`: Për matjen ekzakte të kohës me sa më shumë vlera pas dhjetores.
  * `numpy`: Për gjenerimin e shpejtë të strukturave të të dhënave dhe matricave (input-eve).
  * `matplotlib.pyplot`: Për vizualizimin grafik të rezultateve eksperimentale.

## 3. Metodologjia Eksperimentale

Eksperimenti kompjuterik ndjek pesë hapa kronologjikë:

* **Hapi 1 — Zgjedhja e algoritmit:** Përcaktohet funksioni që do të testohet (p.sh., një cikël që llogarit shumën e numrave).
* **Hapi 2 — Gjenerimi i të dhënave:** Krijohen input-e me madhësi progresive (p.sh., masivë me $n = 100$, $n = 1000$, deri në $n = 10000$ elementë).
* **Hapi 3 — Matja e kohës:** Regjistrohet koha ekzakte përpara fillimit dhe pas përfundimit të algoritmit.
* **Hapi 4 — Përsëritja:** Për të shmangur luhatjet e rastësishme të procesorit, matja për të njëjtin input përsëritet disa herë dhe nxirret mesatarja.
* **Hapi 5 — Analiza grafike:** Ndërtohet grafiku $T(n)$ për të parë sjelljen e kurbës.

```python
import time
import numpy as np
import matplotlib.pyplot as plt

# Definojmë një algoritëm të thjeshtë (p.sh. llogaritja e një shume)
def algoritmi_test(n):
    shuma = 0
    for i in range(n):
        shuma += i
    return shuma

# Vlerat progresive të input-it (n)
velrat_n = [10000, 50000, 100000, 500000, 1000000, 2000000]
kohet_ekzekutimit = []

# Ekzekutimi i matjeve
for n in velrat_n:
    t_fillim = time.perf_counter()
    
    # Ekzekutojmë algoritmin
    algoritmi_test(n)
    
    t_fund = time.perf_counter()
    t_total = t_fund - t_fillim
    kohet_ekzekutimit.append(t_total)

# Shfaqja e rezultateve në terminal
for i, n in enumerate(velrat_n):
    print(f"Input n = {n:7d} -> Koha T(n) = {kohet_ekzekutimit[i]:.6f} s")
```

## 4. Analiza Grafike dhe Rezultatet e Pritshme

Nga të dhënat e mbledhura në kodin e mësipërm, mund të ndërtojmë grafikun përkatës për të vlerësuar kompleksitetin kohor.

```python
plt.figure(figsize=(8, 5))
plt.plot(velrat_n, kohet_ekzekutimit, marker='o', linestyle='-', color='purple', label='T(n) Eksperimentale')
plt.xlabel('Madhësia e Input-it (n)')
plt.ylabel('Koha e Ekzekutimit (sekonda)')
plt.title('Grafiku i Kohës së Ekzekutimit T(n) vs Madhësia e Input-it n')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.savefig('figs/koha_ekzekutimit.png')
plt.show()
```

### Rezultatet e Pritshme
Grafiku tregon një varësi të pastër **lineare** midis kohës dhe madhësisë së input-it. Kjo do të thotë se nëse dyfishojmë vlerën e $n$, koha e ekzekutimit përafërsisht do të dyfishohet gjithashtu. Ky njihet ndryshe si kompleksiteti kohor $O(n)$.

## 5. Appendix – Detaje mbi Matjen e Kohës

### Çfarë matim realisht?
Kur ekzekutojmë komandën në Python, ne masim ndryshimin total të orës së sistemit:
$$T = t_{\text{fund}} - t_{\text{fillim}}$$

Megjithatë, kjo kohë reale (Wall-clock time) nuk përfshin vetëm operacionet e pastra matematike të kodit tonë. Ajo është një përzierje e:
* Kohës efektive të procesorit (CPU time).
* Vonesave të krijuara nga Sistemi Operativ (kur ndërron detyrat në prapavijë).
* Kohës që i duhet sistemit për të aksesuar qelizat e memories RAM.

### Pse duhet përsëritja?
Pikërisht për shkak të këtyre vonesave të paparashikueshme të Sistemit Operativ, një matje e vetme mund të jetë e pasaktë (mund të ketë një kulm artificial sepse kompjuteri po bënte diçka tjetër në atë sekondë). Përsëritja e matjes disa herë dhe nxjerrja e mesatares na ndihmon të lmojmë këto efekte anësore.

* **Përdorimi i librarisë së gabuar:** Përdorimi i funksionit të vjetër `time.time()` në vend të `time.perf_counter()`, i cili nuk ofron rezolucion të mjaftueshëm për matjen e proceseve që zgjasin në milisekonda.
