## 1. Ideja Kryesore: Pse T(n) nuk është konstante?

Në laboratorin e kaluar (LAB09), ne supozuam se për një madhësi të caktuar input-i $n$, koha e ekzekutimit $T(n)$ është një vlerë fikse. Megjithatë, në realitetin e sistemeve kompjuterike moderne, kjo nuk ndodh.

Nëse ekzekutojmë të njëjtin program, me të njëjtin input, disa herë rresht, do të shohim se koha e ekzekutimit ndryshon lehtësisht midis provave. Pra, koha e ekzekutimit sillet si një **madhësi statistikore** që ndjek një shpërndarje probabilitare dhe jo si një konstante matematike.

$$T(n) \sim \text{Shpërndarje (Zhurmë)}$$

## 2. Modeli Fizik: Pse ekziston variacioni?

Variacioni i kohës (i njohur edhe si *jitter*) ndikohet nga një sërë faktorësh kompleksë brenda arkitekturës së kompjuterit:

* **CPU (Procesori):** Frekuenca e punës ndryshon dinamikisht (Turbo Boost) dhe procesori duhet të ndajë kohën e tij midis qindra proceseve të tjera.
* **Memoria:** Ndodhja e *Cache hits* ose *misses* varet nga gjendja e memories përpara ekzekutimit. Lokalitetit i të dhënave luan një rol kyç.
* **Sistemi Operativ (OS):** *Scheduler* i sistemit operativ mund ta ndërpresë programin tonë në çdo moment për t'i dhënë përparësi një procesi në sfond apo një ndërprerjeje harduerike.
* **I/O dhe Faktorët e Jashtëm:** Aksesi në disk, temperatura e lartë që shkakton ulje të shpejtësisë (Thermal Throttling) dhe menaxhimi i energjisë.

Pra, çdo matje e kohës kompjuterike përmban **zhurmë (noise)** reale.

## 3. Qasja Eksperimentale dhe Metodologjia

Në këtë laborator, ndryshe nga i kaluari, ne nuk ndryshojmë input-in $n$, por e mbajmë atë **fiks** dhe përsërisim matjen shumë herë ($N$ herë) për të mbledhur një kampion kohësh: $T_1, T_2, T_3, \dots, T_N$.

**Hapat:**
1. **Zgjedhja e kodit:** Një funksion që kërkon kohë të mjaftueshme për t'u matur (p.sh. llogaritja e numrave primarë).
2. **Fiksimi i input-it:** Zgjedhim një vlerë $n$ që zgjat rreth 50-100 ms.
3. **Matja e përsëritur:** Ekzekutojmë kodin 1000 herë rresht dhe ruajmë çdo kohë.
4. **Analiza statistikore:** Llogaritim mesataren, devijimin standard dhe vlerat ekstreme (min/max).
5. **Vizualizimi:** Ndërtojmë një histogram për të parë formën e zhurmës.

```python
import time
import numpy as np
import matplotlib.pyplot as plt

# Funksioni testues
def operacion_matematik(n):
    return sum(i**2 for i in range(n))

input_fiks = 50000
perseritje = 1000
kohet = []

# Matja e përsëritur
for i in range(perseritje):
    t_start = time.perf_counter()
    operacion_matematik(input_fiks)
    t_end = time.perf_counter()
    kohet.append((t_end - t_start) * 1000) # Konvertim në milisekonda

kohet = np.array(kohet)

print(f"Analiza Statistikore (n={input_fiks}):")
print(f"Mesatarja: {np.mean(kohet):.4f} ms")
print(f"Minimumi:  {np.min(kohet):.4f} ms")
print(f"Maximumi:  {np.max(kohet):.4f} ms")
print(f"Devijimi Standard: {np.std(kohet):.4f} ms")
```

## 4. Vizualizimi i Rezultateve

```python
plt.figure(figsize=(10, 5))

# Histogrami i kohëve
plt.hist(kohet, bins=50, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(np.mean(kohet), color='red', linestyle='dashed', linewidth=2, label=f'Mesatarja: {np.mean(kohet):.2f}ms')

plt.title('Shpërndarja e Kohës së Ekzekutimit (Zhurma e Sistemit)')
plt.xlabel('Koha (ms)')
plt.ylabel('Frekuenca')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.savefig('figs/variacioni_performances.png')
plt.show()
```
