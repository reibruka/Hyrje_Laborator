## 1. Hyrje në Sistemin GPS

GPS (Global Positioning System) është një sistem navigimi satelitor që lejon përcaktimin e pozicionit tredimensional (gjerësi gjeografike, gjatësi gjeografike dhe lartësi) dhe kohës në çdo pikë të Tokës.

### Si funksionon?
Sistemi bazohet në parimin e **Trilateracionit**. Një marrës GPS mat kohën që i duhet sinjalit të udhëtojë nga të paktën katër satelitë deri te marrësi. Duke ditur shpejtësinë e dritës ($c \approx 3 \times 10^8$ m/s), llogaritet distanca deri te çdo satelit:
$$d = c \cdot \Delta t$$

Pozicioni i marrësit gjendet në pikëprerjen e sferave të formuara nga këto distanca.

## 2. Burimet e Pasigurisë në GPS

Edhe pse GPS është shumë i saktë, matjet kanë një pasiguri të natyrshme (zhurmë) që vjen nga:
1. **Vonesat Atmosferike:** Sinjali ngadalësohet gjatë kalimit në Jonosferë dhe Troposferë.
2. **Gabimet e Orës:** Diferenca e vogël midis orëve atomike të satelitëve dhe orës së thjeshtë të marrësit.
3. **Multi-path:** Reflektimi i sinjalit nga ndërtesat ose pengesat përpara se të mbërrijë te marrësi.
4. **Gjeometria e Satelitëve (DOP):** Pozicioni i satelitëve në qiell ndikon në saktësinë e llogaritjes.

## 3. Eksperimenti: Variacioni Statisikor i Pozicionit

Në këtë laborator, ne analizojmë të dhënat e mbledhura nga një marrës GPS statik. Edhe pse marrësi nuk lëviz, koordinatat e raportuara luhaten rreth një pike mesatare. Kjo luhatje përfaqëson "zhurmën" e sistemit.

```python
import numpy as np
import matplotlib.pyplot as plt

# Simulimi i të dhënave GPS statike (pika rreth një koordinatë mesatare)
np.random.seed(42)
n_points = 500
lat_mesatare = 41.3275  # Tirana
lon_mesatare = 19.8187

# Shtojmë zhurmë Gaussian (në gradë, shumë e vogël)
lat_zhurma = np.random.normal(0, 0.00005, n_points)
lon_zhurma = np.random.normal(0, 0.00007, n_points)

latitudes = lat_mesatare + lat_zhurma
longitudes = lon_mesatare + lon_zhurma

# Vizualizimi i shpërndarjes së pikave (Scatter Plot)
plt.figure(figsize=(8, 8))
plt.scatter(longitudes, latitudes, alpha=0.5, s=10, color='blue', label='Pikat e regjistruara')
plt.scatter(lon_mesatare, lat_mesatare, color='red', marker='+', s=100, label='Pozicioni Real (Mesatar)')

plt.title('Variacioni i Pozicionit GPS Statik')
plt.xlabel('Gjatësia Gjeografike (Longitude)')
plt.ylabel('Gjerësia Gjeografike (Latitude)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.savefig('figs/variacioni_gps.png')
plt.show()
```

## 4. Analiza e Gabimit dhe CEP

Për të matur saktësinë e një marrësi GPS, përdoret shpesh **Circular Error Probable (CEP)**. CEP përcakton rrezen e rrethit, të qendërzuar në pozicionin mesatar, brenda të cilit ndodhen 50% e matjeve.

```python
# Llogaritja e distancës nga qendra për çdo pikë
distancat = np.sqrt((latitudes - lat_mesatare)**2 + (longitudes - lon_mesatare)**2)
cep_50 = np.percentile(distancat, 50)

print(f"Devijimi Standard (Lat): {np.std(latitudes):.7f}")
print(f"Devijimi Standard (Lon): {np.std(longitudes):.7f}")
print(f"Rrezja CEP (50%): {cep_50:.7f} gradë gjeografike")
```
